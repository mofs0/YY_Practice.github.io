# 循环

```python
score = int(input('请输入一个分数:'))
if 100 >= score >= 90:
    print('A')
if 90 > score >=80:
    print('B')
if 80 > score >= 70:
    print('C')
if 70 > score >=60:
    print('D')
if 60 > score >=0 :
    print('不及格')
if 0 > score or score > 100:
    print('请输入正确的分数')
```

```python
score = int(input('请输入一个分数:'))

if score < 0 or score > 100:
    print('请输入正确的分数')
else:
    if score >= 90:
        print('A')
    else:
        if score >= 80:
            print('B')
        else:
            if score >= 70:
                print('C')
            else:
                if score >= 60:
                    print('D')
                else:
                    print('不合格')
```

嵌套

```python
score = int(input('请输入一个分数:'))

if 100 >= score >= 90:
    print('A')
elif 90 > score >= 80:
    print('B')
elif 80 > score >= 70:
    print('C')
elif 70 > score >= 60:
    print('D')
elif 0 <= score < 60:
    print('不合格')
else:
    print('请输入正确的分数')
```

break语句

```python
bingo = '清蒸'
answer = input('清蒸还是炖了?')
while True:
    if answer == bingo:
        break
    answer = input('抱歉，输入错误，请重新输入')
print('清蒸好')
```

打印2018年以后出现的第一个闰年

```python
for year in range(2018,200000):
    if (year % 4 == 0 ) & (year % 100 != 0) or (year % 400 ==0):
        break
print('2018年一后出现的第一个闰年是:',year)
```

```python
def num():
    x = input('请输入一个数: ')
    # 检查输入是否为纯数字
    if not x.isdigit():
        print("输入错误，请输入有效的数字！")
        return
    
    # 检查数字是否为三位数及以上
    if len(x) >= 3:
        print(f"去掉最后两位后的结果: {x[:-2]}")
    else:
        print("请输入三位数以上的数字！")
```

```python
num()
```

```python
def ifm():
    x = input('请输入姓名: ')
    
    # 处理性别输入
    y = input('请输入性别(男/女): ')
    # 判断是否为汉字且长度为1
    if len(y) != 1 or not '\u4e00' <= y <= '\u9fff':
        print('性别请输入单个汉字(男/女)')
        return  # 输入错误则结束函数
    
    # 处理年龄输入
    z = input('请输入年龄: ')
    if not z.isdigit() or int(z) <= 0 or int(z) > 150:
        print('请输入有效的年龄数字(1-150)')
        return  # 输入错误则结束函数
    
    # 所有输入验证通过，打印信息
    print('=' * 30)
    print(f'姓名: {x}')
    print(f'性别: {y}')
    print(f'年龄: {z}')
    print('=' * 30)
```

```python
ifm()
```

```python
def cal():
    x = input('请输入加法式（例如：12+34+56）：')
    # 按 '+' 分割成字符串列表
    parts = x.split('+')
    
    try:
        y = 0  # 在循环外初始化累加变量
        # 遍历每个数字字符串
        for part in parts:
            y += int(part)  # 将字符串转换为整数并累加
        return y
    except ValueError:
        print("输入格式错误，请输入正确的加法表达式")
        return None

# 测试
print(cal())  # 例如输入 "1+2+3" 会输出 6
```

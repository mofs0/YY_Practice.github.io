## 创建和使用类

```python
#创建类
class Student:
    '''
    简单的学生类
    '''
    def __init__(self,name,age):
        '''初始化'''
        self.name = name
        self.age = age
    def run(self):
        '''模拟跑步'''
        print(f'{self.name}正在跑步')
    def greet(self):
        '''模拟打招呼'''
        print(f'{self.name}正在打招呼')
        
```

```python
#使用类
if __name__ == '__main__':
    st1 = Student('zhao',19)
    print(f'已经创建了一个实例名为{st1.name}学生')
    st1.run()
```

```python
st2 = Student('xy',20)
print(f'创建了第二个实例对象{st2.name}学生')
st2.greet()
```

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        self.credit=0
    def get_credit(self):
        return self.credit
    def add_credit(self,increment):
        self.credit = self.credit + increment
st1 = Student('yy',15)
if __name__ == '__main__':
    print(f'{st1.name}的学分为{st1.get_credit()}')
    st1.add_credit(5)
    print(f'{st1.name}的学分为{st1.get_credit()}')
```

```python
#私有属性和私有方法
class Cup:
    def __init__(self,capacity):
        self.__capacity = capacity
if __name__ == '__main__':
    cup = Cup(3000)
    print(cup.__capacity)
```

```python
#需要用特殊的方法调用私有属性
class Cup:
    def __init__(self,capacity):
        self.__capacity = capacity
    def get_capacity(self):
        return f'{self.__capacity}毫升'
if __name__ == '__main__':
    cup = Cup(3000)
    print(cup.get_capacity())       
```

## 类属性和类方法

```python
class Person:
    count = 0
    def __init__(self):
        Person.count = Person.count + 1
if __name__ == '__main__':
    zhang = Person()
    print(Person.count)
    print(zhang.count)
    wang = Person()
    print(Person.count)
    print(wang.count)
    print(zhang.count)
```

```python
#析构函数
class Person:
    count = 0
    def __init__(self):
        Person.count = Person.count +1
    def __del__(self):
        Person.count = Person.count -1
def func():
    x = Person()
    print('函数内部:',Person.count)
if __name__ == '__main__':
    wang = Person()
    print('创建一个实例后:',Person.count)
    del wang
    print('释放一个实例后:',Person.count)
    func()
    print('函数外部:',Person.count)
```

```python
#类方法
class Person:
    count = 0
    def __init__(self):
        Person.count = Person.count +1
    def __del__(self):
        Person.count = Person.count-1
    @classmethod
    def get_count(cls):
        return cls.count
if __name__ == '__main__':
    print(Person.get_count())
    zhang = Person()
    print(zhang.get_count())
```

```python
#静态方法
class Person:
    count = 0
    def __init__(self):
        Person.count = Person.count +1
    def __del__(self):
        Person.count = Person.count-1
    @staticmethod
    def get_count():
        return Person.count
if __name__ == '__main__':
    print(Person.get_count())
    zhang = Person()
    print(zhang.get_count())
```

```python
#@property
class Square:
    '''
    正方形类,允许直接读取/设置对象的面积
    该类通过边长计算面积，也可以通过面积反推边长
    '''
    def __init__(self, side):
        # 初始化方法，接收正方形的边长作为参数
        self.side = side  # 存储正方形的边长
    
    @property
    def area(self):
        # 面积属性的getter方法
        # 通过边长计算并返回正方形的面积（边长×边长）
        return self.side * self.side
    
    @area.setter
    def area(self, value):
        # 面积属性的setter方法
        # 接收面积值，通过开平方计算并设置对应的边长
        self.side = pow(value, 0.5)  # pow(value, 0.5) 等价于求平方根

if __name__ == '__main__':
    # 当模块作为主程序运行时执行以下代码
    s1 = Square(2)  # 创建一个边长为2的正方形实例
    print(s1.area)  # 输出正方形的面积（2×2=4）
    s1.area = 9     # 通过设置面积为9，自动计算出对应的边长（3）
    print(s1.side)  # 输出更新后的边长（3）
```

### 继承

```python
#简单的继承
class Student:
    '''学生类'''
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def run(self):
        print(f'{self.name}在跑步')
class SpecialStudent(Student):
    '''特长生类'''
    pass
if __name__ == '__main__':
    st1 = SpecialStudent('zhang',19)
    st1.run()
```

```python
#重写父类的方法
class Student:
    '''学生类'''
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def run(self):
        print(f'{self.name}在跑步')
class SpecialStudent(Student):
    '''特长生类'''
    def run(self):
        print(f'{self.name}跑的更快')
if __name__ == '__main__':
    st1 = SpecialStudent('zhang',19)
    st1.run()
```

```python
#重写__init__方法
class SpecialStudent(Student):
    '''特长生类'''
    def __init__(self,name,age):
        super().__init__(name,age)
        print(f'{self.name}是特长生')
    def run(self):
        super().run()
        print(f'{self.name}跑的更快')
if __name__ == '__main__':
    st1 = SpecialStudent('zhang',19)
    st1.run()
```

```python
#为子类增加新的属性和方法
class SpecialStudent(Student):
    '''特长生类'''
    def __init__(self,special,name,age):
        self.special = special
        super().__init__(name,age)
        print(f'{self.name}是特长生')
    def run(self):
        super().run()
        print(f'{self.name}跑的更快')
    def introduce(self):
        print(f'{self.name}的特长是{self.special}')
if __name__ == '__main__':
    st1 = SpecialStudent('体育','zhang',19)
    st1.run()
    st1.introduce()
```

### 抽象类和抽象方法

```python
from abc import ABC,abstractclassmethod
class Animal(ABC):
    @abstractclassmethod
    def eat(self):
        pass
class Pig(Animal):
    def eat(self):
        print('Pig is eating')
class Dog(Animal):
    def eat(self):
        print('Dog is eating')
pig = Pig()
dog = Dog()
pig.eat()
dog.eat()
```

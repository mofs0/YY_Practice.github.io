```python
#1
class RomanNumeral:
    """
    一个将阿拉伯数字转换为罗马数字的类
    
    用法示例:
    >>> num = RomanNumeral(1994)
    >>> print(num)  # 输出: MCMXCIV
    """
    
    # 定义罗马数字与阿拉伯数字的对应关系（从大到小排列）
    _roman_values = [
        (1000, 'M'),
        (900, 'CM'),
        (500, 'D'),
        (400, 'CD'),
        (100, 'C'),
        (90, 'XC'),
        (50, 'L'),
        (40, 'XL'),
        (10, 'X'),
        (9, 'IX'),
        (5, 'V'),
        (4, 'IV'),
        (1, 'I')
    ]
    
    def __init__(self, number):
        """
        初始化罗马数字转换器
        
        参数:
            number (int): 要转换的阿拉伯数字，必须在1到3999之间（罗马数字的有效范围）
        
        异常:
            ValueError: 如果输入的数字不在有效范围内
            TypeError: 如果输入的不是整数
        """
        # 验证输入类型
        if not isinstance(number, int):
            raise TypeError("输入必须是整数")
        
        # 验证输入范围（罗马数字通常用于表示1到3999之间的数字）
        if number < 1 or number > 3999:
            raise ValueError("罗马数字仅支持1到3999之间的数字")
        
        self.number = number
        self.roman_numeral = self._convert_to_roman()
    
    def _convert_to_roman(self):
        """将阿拉伯数字转换为罗马数字的内部方法"""
        remaining = self.number
        roman = []
        
        # 从最大的数值开始，逐步减去并拼接对应的罗马数字
        for value, symbol in self._roman_values:
            while remaining >= value:
                roman.append(symbol)
                remaining -= value
            if remaining == 0:
                break
        
        return ''.join(roman)
    
    def __str__(self):
        """返回罗马数字的字符串表示"""
        return self.roman_numeral
    
    def __repr__(self):
        """返回类实例的正式表示"""
        return f"RomanNumeral({self.number}) -> '{self.roman_numeral}'"


# 示例用法
if __name__ == "__main__":
    try:
        # 测试一些典型数字
        numbers = [1,2, 3, 4, 9, 58, 1994, 3999]
        for num in numbers:
            roman = RomanNumeral(num)
            print(f"{num} -> {roman}")
    except (ValueError, TypeError) as e:
        print(f"错误: {e}")
```

```python
import math  # 缺少math模块的导入

class Round:
    def __init__(self, r):
        self.r = r
    
    @property
    def area(self):
        return self.r * self.r * math.pi
    
    @area.setter
    def area(self, value):
        self.r = pow(value / math.pi, 0.5)

if __name__ == '__main__':
    s1 = Round(3)
    print(s1.area)  # 输出圆的面积
    
    # s2未实例化，需要先创建对象
    s2 = Round(0)  # 初始半径可以设为0
    s2.area = 10   # 通过面积设置半径
    print(s2.r)    # 输出根据面积计算出的半径
```

```python
class ProductSales:
    """商品销售类，用于记录商品销售信息和计算销售额"""
    
    def __init__(self, retail_price, wholesale_discount=0.2, wholesale_min_qty=10):
        """
        初始化商品销售信息
        
        参数:
            retail_price (float): 商品零售单价
            wholesale_discount (float): 批发折扣百分比，默认20%（0.2）
            wholesale_min_qty (int): 起批发数量，默认10件
        """
        self.retail_price = retail_price  # 零售价格
        self.wholesale_discount = wholesale_discount  # 批发折扣百分比
        self.wholesale_min_qty = wholesale_min_qty    # 起批发数量
        self.total_sales_qty = 0  # 总销售数量
        self.total_revenue = 0.0  # 总销售额
    
    def record_sale(self, quantity):
        """
        记录一次商品销售
        
        参数:
            quantity (int): 销售数量，必须为正整数
            
        返回:
            float: 本次销售额
            
        异常:
            ValueError: 如果销售数量不是正整数
        """
        if not isinstance(quantity, int) or quantity <= 0:
            raise ValueError("销售数量必须是正整数")
        
        # 根据销售数量判断适用价格（零售或批发）
        if quantity >= self.wholesale_min_qty:
            # 批发价 = 零售价 * (1 - 折扣百分比)
            unit_price = self.retail_price * (1 - self.wholesale_discount)
        else:
            unit_price = self.retail_price
        
        # 计算本次销售额
        sales_amount = quantity * unit_price
        
        # 更新总销售数据
        self.total_sales_qty += quantity
        self.total_revenue += sales_amount
        
        return sales_amount
    
    def get_total_sales(self):
        """返回总销售数量和总销售额"""
        return {
            "total_quantity": self.total_sales_qty,
            "total_revenue": round(self.total_revenue, 2)
        }
    
    def __str__(self):
        """返回商品销售信息的字符串表示"""
        return (f"商品销售信息:\n"
                f"零售单价: ¥{self.retail_price:.2f}\n"
                f"批发折扣: {self.wholesale_discount*100:.0f}%\n"
                f"起批数量: {self.wholesale_min_qty}件\n"
                f"总销售量: {self.total_sales_qty}件\n"
                f"总销售额: ¥{self.total_revenue:.2f}")


# 示例用法
if __name__ == "__main__":
    # 创建一个商品销售实例：单价100元，批发折扣30%，起批数量5件
    product = ProductSales(retail_price=100, wholesale_discount=0.3, wholesale_min_qty=5)
    
    # 记录几次销售
    try:
        sale1 = product.record_sale(3)  # 零售
        print(f"销售3件，销售额: ¥{sale1:.2f}")
        
        sale2 = product.record_sale(6)  # 批发
        print(f"销售6件，销售额: ¥{sale2:.2f}")
        
        sale3 = product.record_sale(10) # 批发
        print(f"销售10件，销售额: ¥{sale3:.2f}")
        
        # 查看总销售数据
        total = product.get_total_sales()
        print(f"\n累计销售: {total['total_quantity']}件，累计销售额: ¥{total['total_revenue']:.2f}")
        
        # 打印完整信息
        print("\n" + str(product))
    except ValueError as e:
        print(f"销售记录失败: {e}")
    
```

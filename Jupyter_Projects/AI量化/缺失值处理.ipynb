# 缺失值处理

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(data=np.random.randint(0,100,(8,6)))
df.iloc[2,3]=None
df.iloc[4,4]=None
df.iloc[5,2]=None
df
```

方法一:对空值进行过滤（删除空值所在的行数据)

```python
df.isnull()
```

```python
df.isnull().any(axis=1)#检测每个行是否有空值
df.loc[df.isnull().any(axis=1)]#缺失行的行数据
drop_index = df.loc[df.isnull().any(axis=1)].index#缺失值行的行索引
df.drop(drop_index)
```

```python
df.notnull().all(axis=1)
```

方法二：dropna

```python
df.dropna(axis=0)
```

# 对缺失值进行覆盖

```python
df.fillna(method='ffill',axis=1)#水平方向向前填充
```

```python
df.fillna(method='bfill',axis=1)#水平方向向后填充
```

# 重复值处理

```python
df = pd.DataFrame(data=np.random.randint(0,100,(8,4)))
df.iloc[2]=[0,0,0,0]
df.iloc[4]=[0,0,0,0]
df.iloc[5]=[0,0,0,0]
```

```python
#使用drop_duplicates(keep='first')
df.drop_duplicates(keep='first')
```

# 异常值的处理

```python
df = pd.DataFrame(data=np.random.random((1000,3)),columns=['A','B','C'])
twice_std = df['C'].std()*2#异常值判断条件
df.loc[~(df['C']>twice_std)#取反]
```

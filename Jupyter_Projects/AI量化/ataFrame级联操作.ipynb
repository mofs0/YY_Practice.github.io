# 级联操作

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame(data=np.random.randint(0,100,(5,3)),columns=['a','b','c'])
df2 = pd.DataFrame(data=np.random.randint(0,100,(5,3)),columns=['a','d','c'])
```

```python
pd.concat((df1,df2),axis=1)#按列相连
```

```python
pd.concat((df1,df2),axis=0)#外连接显示所有的数据
```

```python
pd.concat((df1,df2),axis=0,join='inner')#内连接消除了有空值的列
```

```python
df1._append(df2)#按行相连与外连接一致
```

# 合并操作

```python
df1 = pd.DataFrame({'employee':['Lisa','BOb','Jake'],
                   'group':['Accounting','Engineering','Engineering']})
df1
```

```python
df2 = pd.DataFrame({'employee':['Lisa','BOb','Jake'],
                   'hire_date':[2004,2008,2012]})
df2
```

```python
pd.merge(df1,df2,on='employee')
```

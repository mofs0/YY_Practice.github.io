# 映射


```python
dic = {
    'name':['张三','李四','张三'],
    'salary':[15000,20000,15000]
}
df = pd.DataFrame(data=dic)
df
```

```python
#映射关系表
dic = {
    '张三':'tom',
    '李四':'jack'
}
df['e_name']=df['name'].map(dic)
df
```

```python
def after_sal(s):
    return s - (s-3000)*0.5
```

```python
df['after_sal'] = df['salary'].map(after_sal)#充当运算工具
df
```

# 排序实现的随机抽样

```python
df = pd.DataFrame(data=np.random.randint(0,100,(100,3)),columns=['A','B','C'])
df
```

```python
#生成随机的乱序
np.random.permutation(10)
```

```python
#将原始数据打乱
df.take(np.random.permutation(3),axis=1).take(np.random.permutation(100),axis=0)[0:50]
```

# 数据分类处理--分组聚合

```python
df = pd.DataFrame({'item':['Apple','Banana','Orange','Banana','Orange','Apple'],
                   'price':[4,3,3,2.5,4,2],
                   'color':['red','yellow','yellow','green','green','green'],
                   'weight':[12,20,50,30,20,44]
})
df
```

```python
df.groupby(by='item')['price'].mean()
```

```python
df.groupby(by='item')['weight'].mean()
```

```python
dic = df.groupby(by='color')['weight'].mean().to_dict()
df['mean_col'] = df['color'].map(dic)
df
```

```python
v[0] = 'Python'
m
```

### 使用copy的deepcopy可以避免

```python
from copy import deepcopy
v = [0.5,0.75,1.0,1.5,2.0]
m = 3 * [deepcopy(v),]
m
```

```python
v[0] = 'Python'
m
```

### np.where()

```python
#h中大于7的元素换为1否则为0
np.where(h>7,1,0)
```

```python
#奇数为odd偶数为even
np.where(h% 2 == 0 ,'even','odd')
```

```python
#大于7元素值翻倍,否则减半
np.where(h >7,h * 2 ,h/2)
```

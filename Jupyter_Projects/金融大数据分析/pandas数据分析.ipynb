# DataFrame()函数参数

data 数据/字典/ndarray/列表 

index 索引/类似数组 默认为顺序数

columns 索引/类似数组 列标题

copy 布尔值 默认None 从输入复制数据

## 为了高效地处理金融时间序列数据,需要很好的处理时间索引.假设分为四列九行的数据对应于2019年1月开始的月底数据.然后,用date_range()生成一个Datatime对象

```python
#生成时间索引
dates = pd.date_range('2019-1-1',periods=9,freq='M')
dates
```

# date_range()函数的参数

start 字符串/日期时间 生成起始日期

end 字符串/日期时间 生成结束日期

period 整数 期数(生成多少个)

freq 字符串/日期偏移 频率字符串(例如5D,2M)2天,五月

```python
df.index =dates
df
```

```python
#方法二
num = int(input('请输入一个数字'))
def Dec_to_Oct(num):
    result = ""
    while num != 0:
        ret = num % 8
        num //= 8
        result = str(ret) + result
    return result


print('您输入的10进制数其8进制对应的数字为:',Dec_to_Oct(num))
```

第二题

```python
def longest_repeated_substring(s):
    n = len(s)
    longest_substring = ""

    for i in range(2, n + 1, 2):  # 从长度为2的子串开始，逐渐增加长度
        for j in range(n - i + 1):
            substring = s[j:j + i]
            if substring[:i // 2] == substring[i // 2:]:
                if len(substring) > len(longest_substring):
                    longest_substring = substring

    return longest_substring

# 测试示例
input_str = "abababcdabcd"
result = longest_repeated_substring(input_str)
print("最长的自重复子串为:", result)
```

第三题

```python
#要计算无向图的连通分量数量，可以使用深度优先搜索（DFS）算法。
#首先，你需要遍历图中的每个节点，对于每个尚未访问过的节点，进行一次深度优先搜索，并标记所有与该节点连通的节点。
#每次搜索完一个连通分量后，将连通分量的数量加一。
def dfs(graph, visited, node):
    visited[node] = True
    for neighbor in range(len(graph[node])):
        if graph[node][neighbor] == 1 and not visited[neighbor]:
            dfs(graph, visited, neighbor)

def count_connected_components(graph):
    n = len(graph)
    visited = [False] * n
    count = 0

    for i in range(n):
        if not visited[i]:
            dfs(graph, visited, i)
            count += 1

    return count

# 输入
n = int(input("请输入图的节点个数："))
print("请输入邻接矩阵A：")
graph = []
for _ in range(n):
    row = list(map(int, input().split()))
    graph.append(row)

# 输出
result = count_connected_components(graph)
print("该图的连通分量个数为:", result)
```

第四题

```python
#这题有点难度自己去查找了相关的资料才做出
from typing import List, Tuple
import operator
from functools import reduce

class Solution:
    def minFlips(self, mat: List[List[int]]) -> Tuple[List[Tuple[int, int]], int]:
        rows, cols = len(mat), len(mat[0])
        start = reduce(operator.ior, [t << i * cols + j for i, row in enumerate(mat) for j, t in enumerate(row)])
        binary_str = bin(start)[2:][::-1]
        flips = []
        flip_count = 0
        for index, bit in enumerate(binary_str):
            if bit == '1':
                row = index // cols
                col = index % cols
                flips.append((row, col))
                flip_count += 1
    
        return flips, flip_count

# 创建 Solution 类的实例
solution = Solution()
mat = [
    [0, 0, 1, 0, 0, 0],
    [0, 1, 1, 1, 0, 1],
    [0, 0, 1, 0, 1, 1]
]
result_positions, result_flip_count = solution.minFlips(mat)
print("需要翻转的位置为:", result_positions)
print("需要翻转的最小次数为:", result_flip_count)
```

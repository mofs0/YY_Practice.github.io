```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression,SGDRegressor
from sklearn.metrics import mean_squared_error
```

```python
def linear1():
    #1.获取数据
    diabetes = load_diabetes()
    print('特征数量: \n',diabetes.data.shape)
    #2.划分数据集
    x_train,x_test,y_train,y_test = train_test_split(diabetes.data,diabetes.target,random_state=22)
    #3.标准化
    transfer = StandardScaler()
    x_train = transfer.fit_transform(x_train)
    x_test = transfer.transform(x_test)
    #4.预估器
    estimator = LinearRegression()
    estimator.fit(x_train,y_train)
    #5.得出模型
    print('正规方程权重系数为: \n',estimator.coef_)
    print('正规方程偏置为: \n',estimator.intercept_)
    #6.模型评估
    y_predict = estimator.predict(x_test)
    error = mean_squared_error(y_test,y_predict)
    print('正规方方程均方误差为: \n',error)
    
    return None
```

```python
def linear2():
    #1.获取数据
    diabetes = load_diabetes()
    #2.划分数据集
    x_train,x_test,y_train,y_test = train_test_split(diabetes.data,diabetes.target,random_state=22)
    #3.标准化
    transfer = StandardScaler()
    x_train = transfer.fit_transform(x_train)
    x_test = transfer.transform(x_test)
    #4.预估器
    estimator = SGDRegressor(learning_rate='constant',eta0=0.1,max_iter=10000)
    estimator.fit(x_train,y_train)
    #5.得出模型
    print('梯度下降权重系数为: \n',estimator.coef_)
    print('梯度下降偏置为: \n',estimator.intercept_)
    #6.模型评估
    y_predict = estimator.predict(x_test)
    error = mean_squared_error(y_test,y_predict)
    print('梯度下降均方误差为: \n',error)
    return None
```

```python
if __name__ == '__main__':
    linear1()
    linear2()
```

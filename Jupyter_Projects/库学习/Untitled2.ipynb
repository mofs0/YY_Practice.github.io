```python
import chardet

# 读取 CSV 文件的内容
with open('data_lianjia_20200423.csv', 'rb') as f:
    # 使用 chardet 检测编码方式
    result = chardet.detect(f.read())

# 输出检测到的编码方式
print("Detected encoding:", result['encoding'])
with open('your_file.csv', 'r', encoding='gb2312', errors='ignore') as f:
    content = f.read()
```

```python
with open('data_lianjia_20200423.csv', 'rb',errors='ignore') as f:
    content = f.read().decode('gb2312')  # 使用原始编码方式解码

# 将文本内容使用新的编码方式重新编码
new_content = content.encode('utf-18')  # 使用 UTF-16 编码方式重新编码

# 将重新编码后的内容写入新的 CSV 文件
with open('new_file.csv', 'wb') as f:
    f.write(new_content)
```

```python
#合并
df1 = pd.DataFrame({'key':['b','b','a','c','a','a','b'],'data1':range(7)})
df2 = pd.DataFrame({'key':['a','b','d'],'data2':range(3)})
df1
```

```python
df3 = pd.DataFrame({'lkey':['b','b','a','c','a','a','b'],'data3':range(7)})
df4 = pd.DataFrame({'rkey':['a','b','d'],'data4':range(3)})
pd.merge(df3,df4,left_on='lkey',right_on='rkey')
#inner只使用两个表中观察到的组合键
#left 使用左表中找到的所有组合键
#right 使用右表中找到的所有组合键
#output 使用两个表观察到的所有组合键
```

```python
data = np.random.randn(30).cumsum()
plt.plot(data,'k--',label = 'Default')
plt.plot(data,'k-',drawstyle = 'steps-post',label = 'steps-post')
plt.legend(loc='best')
```

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)
data = np.random.randn(1000).cumsum()
ax.plot(data,label='realtime data')
ticks = ax.set_xticks([0,250,500,750,1000])
labels = ax.set_xticklabels(['Q1','Q2','Q3','Q4','Q5'],rotation=30,fontsize = 'small')
ax.set_title('My Plot')
ax.set_xlabel('Quarter')
ax.set_ylabel('Price')
ax.annotate('tank',xy=(600,data[600]+10),xytext=(600,data[600]+65),arrowprops=dict(facecolor = 'black',headwidth=4,width=2,headlength=4),horizontalalignment='left',verticalalignment='top')
plt.legend(loc='best')
```

```python
from mpl_toolkits.mplot3d import Axes3D
def fun(x,y):
    return x ** 2 + y ** 2
fig =plt.figure()
ax = fig.add_subplot(111,projection = '3d')
x = y =np.arange(-3,3,0.05)
x,y=np.meshgrid(x,y)
zs = np.array([fun(x,y) for x,y in zip(np.ravel(x),np.ravel(y))])
z = zs.reshape(x.shape)

ax.plot_surface(x,y,z)
ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')
ax.set_zlabel('Z Label')
plt.show()
```

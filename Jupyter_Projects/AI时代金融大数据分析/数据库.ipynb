## JSON数据交换格式

```python
#json数据编码
import json
```

```python
py_dict = {'name':'tony','age':20,'sex':True}
py_list = [1,3]
py_tuple =('A','B','C')#创建元组对象
py_dict['a'] = py_list #添加列表到字典里
py_dict['b'] = py_tuple #添加元组到字典里面
print(py_dict)
print(type(py_dict))
```

```python
#编码
json_obj =  json.dumps(py_dict)
print(json_obj)
print(type(json_obj))
```

```python
json_obj = json.dumps(py_dict,indent=4)
#漂亮的格式化字符串输出
print(json_obj)
```

```python
#写入json到data1.json
with open('data/data1.json','w') as f:
    json.dump(py_dict,f)
#写入及json数据到data2.json
with open('data/data2.json','w') as f:
    json.dump(py_dict,f,indent=4)
```

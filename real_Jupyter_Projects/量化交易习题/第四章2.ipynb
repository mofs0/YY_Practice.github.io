```python
def print_symmetric_diamond(size):
    # 上半部分（包括中间行）
    for i in range(1, size + 1):
        # 计算左侧空格数量，实现居中效果
        spaces = " " * (size - i)
        # 打印空格和星号
        print(spaces + "*" * (2 * i - 1))
    
    # 下半部分（不包括中间行）
    for i in range(size - 1, 0, -1):
        spaces = " " * (size - i)
        print(spaces + "*" * (2 * i - 1))

# 打印大小为5的对称菱形
print_symmetric_diamond(5)
```

```python
num = [2, 7, 11, 15, 1, 8, 7]
pairs = []

# 遍历列表中所有可能的两两组合
for i in range(len(num)):
    for j in range(i + 1, len(num)):  # j从i+1开始，避免重复组合
        if num[i] + num[j] == 9:
            pairs.append((num[i], num[j]))

print("两两相加等于9的元素组合有：")
for pair in pairs:
    print(pair)
```

```python
# 百钱买百鸡问题求解
# 公鸡5元/只，母鸡3元/只，小鸡3只1元
# 要求用100元买100只鸡，且公鸡、母鸡、小鸡都必须有

# 初始化计数器
count = 0

# 遍历所有可能的公鸡数量（最多20只，因为5*20=100）
for rooster in range(1, 20):
    # 遍历所有可能的母鸡数量（最多33只，因为3*33=99）
    for hen in range(1, 33):
        # 计算小鸡数量
        chick = 100 - rooster - hen
        # 确保小鸡数量为正数且能被3整除（因为3只1元）
        if chick > 0 and chick % 3 == 0:
            # 检查总花费是否为100元
            if rooster * 5 + hen * 3 + chick / 3 == 100:
                count += 1
                print(f"解 {count}: 公鸡{rooster}只, 母鸡{hen}只, 小鸡{chick}只")
    
```

```python
def bubble_sort(arr):
    """
    冒泡排序算法实现（升序）
    
    参数:
        arr: 需要排序的列表
        
    返回:
        排序后的列表
    """
    # 获取列表长度
    n = len(arr)
    
    # 外层循环控制需要进行多少轮比较
    for i in range(n):
        # 内层循环控制每轮比较的元素范围
        # 每完成一轮，最大的元素会"冒泡"到末尾，因此下一轮可以少比较一次
        for j in range(0, n-i-1):
            # 如果当前元素大于下一个元素，则交换它们
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    
    return arr

# 测试冒泡排序
if __name__ == "__main__":
    # 测试用例
    test_list = [64, 34, 25, 12, 22, 11, 90]
    print("排序前的列表:", test_list)
    
    # 调用冒泡排序
    sorted_list = bubble_sort(test_list)
    print("排序后的列表:", sorted_list)
```

```python
def is_leap_year(year):
    """判断是否为闰年"""
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)

def day_of_year(year, month, day):
    """计算给定日期是当年的第几天"""
    # 每个月的天数（非闰年）
    days_in_month = [31, 28, 31, 30, 31, 30,
                     31, 31, 30, 31, 30, 31]
    
    # 检查月份和日期的有效性
    if month < 1 or month > 12:
        return "月份无效"
    if day < 1:
        return "日期无效"
    
    # 处理闰年的二月
    if is_leap_year(year):
        days_in_month[1] = 29  # 闰年二月有29天
    
    # 检查日期是否超过当月最大天数
    if day > days_in_month[month - 1]:
        return "日期超过当月最大天数"
    
    # 计算总天数
    total_days = sum(days_in_month[:month - 1]) + day
    return total_days

# 用户输入部分
try:
    year = int(input("请输入年份: "))
    month = int(input("请输入月份: "))
    day = int(input("请输入日期: "))
    
    result = day_of_year(year, month, day)
    if isinstance(result, int):
        print(f"{year}年{month}月{day}日是当年的第{result}天")
    else:
        print(f"错误: {result}")
except ValueError:
    print("错误: 请输入有效的数字")
```

```python
def is_leap_year(year):
    """判断是否为闰年"""
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)

def get_days_in_month(year, month):
    """获取指定年月的天数"""
    if month == 2:
        return 29 if is_leap_year(year) else 28
    # 四月、六月、九月、十一月有30天
    if month in [4, 6, 9, 11]:
        return 30
    return 31

def calculate_first_day(year):
    """计算一年中每个月第一天是星期几（0=星期一, 6=星期日）"""
    first_days = []
    # 1月1日的星期几可以用Zeller公式计算
    # 简化处理：使用已知的基准日期计算，这里以2000年1月1日是星期六(5)为基准
    base_year = 2000
    base_first_day = 5  # 2000年1月1日是星期六
    
    total_days = 0
    for y in range(base_year, year):
        total_days += 366 if is_leap_year(y) else 365
    
    # 计算1月1日的星期几
    jan_first = (base_first_day + total_days) % 7
    first_days.append(jan_first)
    
    # 计算其他月份的第一天
    for month in range(1, 12):
        days = get_days_in_month(year, month)
        next_first = (first_days[-1] + days) % 7
        first_days.append(next_first)
    
    return first_days

def print_calendar(year):
    """打印指定年份的日历"""
    print(f"\n{'='*30}")
    print(f"{' '*10}{year}年 日历")
    print(f"{'='*30}\n")
    
    # 月份名称
    month_names = ["一月", "二月", "三月", "四月", "五月", "六月",
                   "七月", "八月", "九月", "十月", "十一月", "十二月"]
    
    # 获取每个月第一天是星期几
    first_days = calculate_first_day(year)
    
    # 打印每个月的日历
    for month in range(12):
        print(f"[{month_names[month]}]")
        print("一 二 三 四 五 六 日")
        
        days_in_month = get_days_in_month(year, month + 1)
        first_day = first_days[month]
        
        # 打印月初的空格
        print("   " * first_day, end="")
        
        # 打印日期
        for day in range(1, days_in_month + 1):
            # 格式化日期，确保对齐
            print(f"{day:2d} ", end="")
            # 每行7个日期
            if (first_day + day) % 7 == 0:
                print()
        print("\n")

if __name__ == "__main__":
    try:
        year = int(input("请输入年份: "))
        if year <= 0:
            print("请输入有效的年份")
        else:
            print_calendar(year)
    except ValueError:
        print("请输入有效的整数年份")
```

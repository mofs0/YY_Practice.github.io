## 时间与日期模块

```python
#date模块
from datetime import date
```

```python
d = date.today()
d
```

```python
d = date(2020,8,9)
print(d.year)
print(d.month)
print(d.day)
print(d.weekday())
print(d.strftime('%Y/%m/%d'))
```

```python
#time 模块
from datetime import time
```

```python
t = time(20,12,49,883)
print(t)
print(t.hour)
print(t.minute)
print(t.second)
print(t.microsecond)
print(t.strftime('%H:%M:%S.%f'))
```

```python
#datetime类
from datetime import datetime
dt = datetime(2020,8,9,18,22,45,998)
print(datetime.today())
print(datetime.now())
print(dt.year)
print(dt.month)
print(dt.day)
print(d.weekday())
print(dt.strftime('%Y/%m/%d'))
print(dt.hour)
print(dt.minute)
print(dt.second)
print(dt.microsecond)
print(dt.strftime('%H:%M:%S.%f'))
```

```python
# 创建一个列表
my_list = [1, 2, 3]

# append()：在末尾添加元素
my_list.append(4)
print(my_list)  # [1, 2, 3, 4]

# insert()：在指定位置插入元素（索引0处插入0）
my_list.insert(0, 0)
print(my_list)  # [0, 1, 2, 3, 4]

# remove()：移除第一个匹配的元素（移除2）
my_list.remove(2)
print(my_list)  # [0, 1, 3, 4]

# pop()：移除并返回指定索引的元素（默认最后一个）
last_element = my_list.pop()
print(last_element)  # 4
print(my_list)       # [0, 1, 3]
```

## 第四题

```python
def resort():
    x = input('请输入一个字符串')
    y = []
    for i in x:
        y += i
    y.reverse()
    z = ''.join(y)
    print(z)
resort()
```

```python
def resort_1():
    x = input('请输入一个字符串：')
    print(x[::-1])  # 切片 [::-1] 直接反转字符串
resort_1()
```

```python
li = ['alex', 'eric', 'rain', 'Tony', 'Kelly']

# 删除索引2到4的元素（切片是左闭右开，5表示不包含5）
del li[2:5]

print(li)  # 输出: ['alex', 'eric']
```

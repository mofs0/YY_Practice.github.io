## 读取csv

```python
import csv
with open('./stock_data.csv','r') as f:
    f_csv = csv.reader(f)
    for line in f_csv:
        print(line)
```

## 写csv文件

```python
# 关键：恢复Python内置的open()函数（解决被覆盖的问题）
import builtins
open = builtins.open  # 将open重新指向内置函数

# 然后再运行完整的股票数据获取和写入代码
import csv
import tushare as ts

# 1. 初始化Tushare
ts.set_token('598eb3c437b4143dde41a08f283e909b5cce0154bb923c346d4c6e58')
pro = ts.pro_api()

# 2. 获取数据
df = pro.daily(ts_code='600000.SH', start_date='20200101', end_date='20200108')

# 3. 数据预处理
df = df.sort_values(by='trade_date').set_index('trade_date')  # 链式调用更简洁
df = df[['open', 'close', 'high', 'low']]

# 4. 写入CSV（此时open()已恢复为内置函数，不会报错）
with open(r'./stock_data.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['date', 'open', 'close', 'high', 'low'])  # 表头
    for date, row in df.iterrows():
        # 提取数据（用明确的列名，避免歧义）
        row_data = [date, row['open'], row['close'], row['high'], row['low']]
        writer.writerow(row_data)

print("CSV文件已保存！")
# 验证：读取CSV确认结果
import pandas as pd
print("\nCSV文件内容：")
print(pd.read_csv('./stock_data.csv'))
```

# SQl数据库

## 连接,关闭数据库

```python
import sqlite3
conn = sqlite3.connect('test.db')
if conn:
    print('连接数据库成功')
conn.close()
```

## 创建表

```python
import sqlite3
conn = sqlite3.connect('test.db')
c = conn.cursor()
c.execute('''CREATE TABLE EMP
       (ID INT PRIMSRY KEY   NOT NULL,
        NAME           TEXT  NOT NULL,
        AGE            INT   NOT NULL,
        SALARY         REAL)''')
conn.close()
```

## 插入数据

```python
import sqlite3

# 连接到SQLite数据库（如果不存在则创建）
conn = sqlite3.connect('test.db')

# 创建一个游标对象
c = conn.cursor()

# 先创建表（如果不存在）
c.execute('''CREATE TABLE IF NOT EXISTS EMP
             (ID INT PRIMARY KEY     NOT NULL,
             NAME           TEXT    NOT NULL,
             AGE            INT     NOT NULL,
             SALARY         REAL);''')

# 插入数据，使用双引号包裹字符串值
c.execute('INSERT INTO EMP (ID,NAME,AGE,SALARY) \
            VALUES(1,"ZHANG",28,160000)');
c.execute('INSERT INTO EMP (ID,NAME,AGE,SALARY)\
            VALUES(2,"ZHANG",28,160000)');
c.execute('INSERT INTO EMP (ID,NAME,AGE,SALARY)\
            VALUES(3,"ZHANG",28,160000)');

# 提交事务
conn.commit()

# 验证插入结果
c.execute("SELECT * FROM EMP")
print("插入的记录:")
for row in c.fetchall():
    print(row)

# 关闭数据库连接
conn.close()
    
```

## 查询数据

```python
import sqlite3
conn = sqlite3.connect('test.db')
c = conn.cursor()
cursor = c.execute('SELECT * from EMP')
for row in cursor:
    print(row[1],row[2],row[3],sep='')
```

## 修改数据

```python
import sqlite3

# 连接到数据库
conn = sqlite3.connect('test.db')
c = conn.cursor()

# 更新数据 - 使用双引号包裹SQL中的字符串，避免与Python的单引号冲突
c.execute('UPDATE EMP set SALARY = 20000 where NAME = "ZHANG"')
conn.commit()  # 提交更新

# 查询更新后的数据
# 同样使用双引号包裹查询条件中的字符串
cursor = conn.execute('SELECT * from EMP where NAME = "ZHANG"')

# 打印查询结果
print("更新后的员工信息（姓名 年龄 工资）：")
for row in cursor:
    # 打印姓名、年龄、工资，用空格分隔
    print(row[1], row[2], row[3], sep=' ')

# 关闭数据库连接
conn.close()
    
```

## 删除数据

```python
import sqlite3
conn = sqlite3.connect('test.db')
c = conn.cursor()
c.execute("DELETE from EMP where ID='2'")
conn.commit()
cursor = c.execute("SELECT * from EMP")
for row in cursor:
    print(row[1],row[2],row[3])
```

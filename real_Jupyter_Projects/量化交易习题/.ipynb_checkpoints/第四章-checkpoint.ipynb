```python
import math  # 导入math模块以使用sin函数

def f(x):
    if x == 0:
        return 1
    else:
        return math.sin(x) / x  # 使用math.sin()调用正弦函数

# 测试函数
print(f(0))       # 输出: 1
print(f(math.pi)) # 输出: sin(π)/π = 0.0
print(f(0.5))     # 输出: sin(0.5)/0.5 的近似值
    
```

## 3

```python
def is_triangle(a, b, c):
    # 首先检查边长是否为正数
    if a <= 0 or b <= 0 or c <= 0:
        return False
    
    # 检查任意两边之和是否大于第三边
    if (a + b > c) and (a + c > b) and (b + c > a):
        return True
    else:
        return False

# 测试函数
print(is_triangle(3, 4, 5))   # 输出: True（直角三角形）
print(is_triangle(1, 2, 3))   # 输出: False（1+2=3，不满足条件）
print(is_triangle(5, 5, 5))   # 输出: True（等边三角形）
print(is_triangle(2, 3, 4))   # 输出: True（普通三角形）
print(is_triangle(-1, 2, 3))  # 输出: False（存在负数边长）
```

## 5

```python
def convert_case(char):
    # 检查输入是否为单个字符
    if len(char) != 1:
        return "请输入单个字符"
    
    # 判断是否为大写字母并转换
    if char.isupper():
        return char.lower()
    # 判断是否为小写字母并转换
    elif char.islower():
        return char.upper()
    # 非字母字符直接返回
    else:
        return "输入的不是字母"

# 测试函数
print(convert_case('A'))  # 输出: 'a'
print(convert_case('b'))  # 输出: 'B'
print(convert_case('3'))  # 输出: '输入的不是字母'
print(convert_case('AB')) # 输出: '请输入单个字符'
    
```

```python
def f_2(x):
    # 检查输入是否大于3
    if x <= 3:
        print('请输入大于3的数字')
        return  # 直接返回，不执行后续判断
    
    # 判断是否为3的倍数（使用取余%，余数为0表示是倍数）
    if x % 3 == 0:
        print('是3的倍数')
    # 判断是否为5的倍数
    elif x % 5 == 0:
        print('是5的倍数')
    # 判断是否为7的倍数
    elif x % 7 == 0:
        print('是7的倍数')
    # 都不是的情况
    else:
        print('不是3、5、7的倍数')

# 测试函数
f_2(2)    # 输出：请输入大于3的数字
f_2(6)    # 输出：是3的倍数
f_2(10)   # 输出：是5的倍数
f_2(14)   # 输出：是7的倍数
f_2(8)    # 输出：不是3、5、7的倍数
```

## 第10题

```python
# 无限循环，使程序可以持续接收用户输入
while True:
    # 提示用户输入四则运算表达式，或输入Q退出程序
    expr = input("请输入四则运算表达式（如5+6，输入Q退出）：")
    
    # 判断用户是否输入Q（不区分大小写），如果是则退出循环，程序结束
    if expr.upper() == 'Q':
        break
    
    # 分别处理加、减、乘、除（整除）四种运算
    # 判断表达式中是否包含加号+
    if '+' in expr:
        # 使用加号分割表达式，得到两个操作数，并转换为整数
        a, b = map(int, expr.split('+'))
        # 计算并输出加法结果
        print(f"结果：{a + b}")
    # 判断表达式中是否包含减号-
    elif '-' in expr:
        # 使用减号分割表达式，得到两个操作数，并转换为整数
        a, b = map(int, expr.split('-'))
        # 计算并输出减法结果
        print(f"结果：{a - b}")
    # 判断表达式中是否包含乘号*
    elif '*' in expr:
        # 使用乘号分割表达式，得到两个操作数，并转换为整数
        a, b = map(int, expr.split('*'))
        # 计算并输出乘法结果
        print(f"结果：{a * b}")
    # 判断表达式中是否包含整除//
    elif '//' in expr:
        # 使用//分割表达式，得到两个操作数，并转换为整数
        a, b = map(int, expr.split('//'))
        # 计算并输出整除结果
        print(f"结果：{a // b}")
    # 如果以上运算符都不包含，说明输入格式错误
    else:
        print("输入格式错误，请确保是合法的四则运算表达式")
```

## 11

```python
# 获取用户输入的正整数，并转换为整数类型
m = int(input("请输入一个正整数："))
# 先假设输入的数是素数，用is_prime变量标记
is_prime = True
# 素数定义是大于1的自然数，所以如果输入的数小于等于1，肯定不是素数
if m <= 1:
    is_prime = False
else:
    # 只需要检查到该数的平方根即可，因为如果有大于平方根的因数，必然有一个小于平方根的因数与之对应
    for i in range(2, int(m**0.5) + 1):
        # 如果m能被i整除，说明m除了1和它本身还有其他因数，不是素数
        if m % i == 0:
            is_prime = False
            # 只要找到一个因数，就可以确定不是素数，直接跳出循环
            break
# 根据is_prime的值，输出对应的结果
if is_prime:
    print(f"{m} is a prime number.")
else:
    print(f"{m} is not a prime number.")
```

```python
# 遍历 100 到 200 之间的每一个整数
for num in range(100, 201):
    # 先假设当前数字 num 是素数，用 is_prime 标记为 True
    is_prime = True
    # 素数定义是大于 1 的自然数，若 num 小于等于 1，肯定不是素数，将 is_prime 标记为 False
    if num <= 1:
        is_prime = False
    else:
        # 只需要检查到 num 的平方根，因为若有大于平方根的因数，必然有一个小于平方根的因数与之对应
        for i in range(2, int(num**0.5) + 1):
            # 若 num 能被 i 整除，说明 num 除了 1 和它本身还有其他因数，不是素数，将 is_prime 标记为 False 并跳出内层循环
            if num % i == 0:
                is_prime = False
                break
    # 若 is_prime 仍为 True，说明 num 是素数，打印该素数，end=" " 表示打印后不换行，以空格分隔
    if is_prime:
        print(num, end=" ")
# 最后打印一个换行，使输出更整洁
print()
```

```python
#15
# 初始化求和变量为0
sum_all = 0
# 定义变量n为50
n = 50 
# 循环遍历2到50的整数（包含50）
for i in range(2, 51):
    # 这里存在逻辑错误：i//2 == 0 只有当i=2时成立（3//2=1，4//2=2，都不满足）
    if i % 2 == 0:
        # 如果条件成立，累加1/50到sum_all
        sum_all += 1/i
# 输出最终的求和结果
print(sum_all)
```

```python
#99乘法表
# 外层循环控制行数，从 1 到 9
for i in range(1, 10):
    # 内层循环控制每行的列数，列数不超过当前行数
    for j in range(1, i + 1):
        # 打印乘法表达式及结果，end=" " 表示打印后不换行，以空格分隔
        print(f"{j}×{i}={i * j}", end=" ")
    # 每行打印完毕后换行
    print()
```

```python
s = input('请输入一个字符串')
digit_count = 0
letter_count = 0
for char in s:
    if char.isdigit():
        digit_count +=1
    elif char.isalpha:
        letter_count +=1
# 打印统计结果（优化换行格式，避免多余逗号）
print(f'字符串中数字的个数：{digit_count}')
print(f'字符串中字母的个数：{letter_count}')
```

```python
# 获取用户输入的字符串
s = input("请输入一个字符串：")

# 创建空字典用于存储字符及其出现次数
char_count = {}

# 遍历字符串中的每个字符
for char in s:
    # 如果字符已在字典中，计数加1
    if char in char_count:
        char_count[char] += 1
    # 如果字符不在字典中，添加到字典并设置计数为1
    else:
        char_count[char] = 1

# 输出统计结果
print("字符串中各个字符的出现次数：")
for char, count in char_count.items():
    print(f"字符 '{char}' 出现了 {count} 次")
```

```python
def digit(n, k):
    # 检查输入是否为数字（先转换为字符串再判断）
    if not str(n).isdigit() or not str(k).isdigit():
        return "请输入数字"
    
    # 将k转换为整数，方便后续计算
    k = int(k)
    # 将数字转换为字符串，便于获取长度和指定位置
    n_str = str(n)
    a = len(n_str)
    
    # 检查k是否有效（位置从1开始计数）
    if k < 1 or k > a:
        return f"k必须在1到{str(a)}之间"
    
    # 返回第k位的数字（字符串索引从0开始，所以k需要减1）
    return int(n_str[k-1])

# 测试函数
print(digit(12345, 3))   # 输出：3（12345的第3位是3）
print(digit(9876, 5))    # 输出：k必须在1到4之间
print(digit("abc", 2))   # 输出：请输入数字
print(digit(123, "x"))   # 输出：请输入数字
```

```python
def digit(n, k):
    # 检查输入是否为数字
    if not str(n).isdigit() or not str(k).isdigit():
        print('请输入数字')
        return  # 用return代替break，退出函数
    
    k = int(k)
    n_str = str(n)
    a = len(n_str)
    
    # 修正逻辑运算符：用and代替|（|是位运算符）
    if k < 1 or k > a:  # 这里应该用or表示"或者"关系
        print(f'k必须在1到{str(a)}之间')
        return  # 用return退出函数
    
    # 返回第k位数字（索引从0开始，所以k-1）
    return int(n_str[k-1])

# 测试
print(digit(12345, 3))   # 输出：3
print(digit(123, 5))     # 输出提示信息并返回None
print(digit("abc", 2))   # 输出提示信息并返回None
print(digit(9876, "2"))  # 输出：8
```

```python
def gcd(a, b):
    """
    计算两个整数的最大公约数（Greatest Common Divisor）
    使用欧几里得算法（辗转相除法）
    """
    # 确保输入为整数
    if not isinstance(a, int) or not isinstance(b, int):
        raise TypeError("输入必须是整数")
    
    # 处理负数（取绝对值，因为公约数与正负无关）
    a, b = abs(a), abs(b)
    
    # 欧几里得算法：gcd(a, b) = gcd(b, a % b)，直到b为0
    while b != 0:
        a, b = b, a % b
    return a

def lcm(a, b):
    """
    计算两个整数的最小公倍数（Least Common Multiple）
    利用公式：lcm(a, b) = |a * b| / gcd(a, b)
    """
    # 调用最大公约数函数计算
    common_divisor = gcd(a, b)
    
    # 处理0的情况（0和任何数的最小公倍数是0）
    if common_divisor == 0:
        return 0
    
    # 应用公式计算最小公倍数
    return abs(a * b) // common_divisor

# 测试函数
if __name__ == "__main__":
    num1 = 24
    num2 = 36
    print(f"{num1}和{num2}的最大公约数是：{gcd(num1, num2)}")  # 输出：12
    print(f"{num1}和{num2}的最小公倍数是：{lcm(num1, num2)}")  # 输出：72
    
    num3 = -15
    num4 = 25
    print(f"{num3}和{num4}的最大公约数是：{gcd(num3, num4)}")  # 输出：5
    print(f"{num3}和{num4}的最小公倍数是：{lcm(num3, num4)}")  # 输出：75
```

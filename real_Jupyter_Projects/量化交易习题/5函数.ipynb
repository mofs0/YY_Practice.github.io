```python
#形参
def swap(x,y):
    temp = x
    x = y
    y = temp
    print(f'函数内部:x={x},y={y}')
a,b = 1,2
print(f'{a},{b}')
swap(a,b)
print(f'{a},{b}')
```

```python
def f(a,li=[]):
    li.append(a)
    return li
f(1)
```

```python
f(2)
```

```python
def f(a,li=None):
    if li is None:
        li = []
    li.append(a)
    return li
f(1)
```

```python
f(2)
```

```python
#关键字参数
def f(a,b=10,c=20):
    print(a,b,c)
#正确调用方式
f(1,2,3)
f(1,c=2)
f(a=1)
f(b=1,a=3,c=2)
```

```python
#错误调用
f(a=1,2) #关键字参数之后不能出现非关键字参数
f(1,a=1) #重复指定a
f(b=2) #参数a未定
f(1,d=3)#没有参数d
```

```python
#元组参数的封装宇拆封
def func(*name):
    print(type(name))
    print(name)
func(1,2,3)
func(1,2,3,4)
```

```python
def func(a,b=2,*name):
    print(type(name))
    print(name)
func(1,2,3)
func(1,2,3,4)
```

```python
def func(a,b=2,*name,last_para):
    print(type(name))
    print(name)
    print(last_para)
func(1,2,3,last_para='ok')
```

```python
#元组的拆封
def func(a,b,c):
    print(a,b,c)
args = (1,2,3)
func(*args)
```

```python
#字典的封装
def func(**dict):
    print(type(dict))
    print(dict)
func(a=1,b=2,c=3)
```

```python
#解封
def func(a,b,c):
    print(a,b,c)
args = {'a':1,'b':2,'c':3}
func(**args)
```

```python
#函数嵌套
x = 100
def outer():
    def inner():
        nonlocal x
        x = x +10
    x = 50
    inner()
    print(x)
```

```python
outer()
print(x)
```

# 异常处理

```python
while True:
    try:
        x = int(input('请输入一个数'))
        break
    except ValueError:
        print('输入的不是整数,重新输入')
```

```python
while True:
    try:
        x = int(input('请输入一个数'))
        y  = 10/x
        break
    except ValueError:
        print('输入的不是整数,重新输入')
    except ZeroDivisionError:
        print('0不能做除数')
print(y)   
```

```python
while True:
    try:
        x = int(input('请输入一个数'))
        y  = 10/x
        break
    except ZeroDivisionError:
        print('0不能做除数')
    except BaseException as ex:
        print(type(ex))
        print(ex.args)
print(y)   
```

```python
while True:
    try:
        x = int(input('请输入一个数'))
        y = 10 /x
    except ZeroDivisionError:
        print('除数不能为0,请重新输入')
    except BaseException as ex:
        print(type(ex))
        print(ex.args)
    else:
        print('除数为:',x)
        break
print(y)
```

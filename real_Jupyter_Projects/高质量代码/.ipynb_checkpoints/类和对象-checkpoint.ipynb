# 对象 = 属性 + 方法

```python
class Ball:
    def setName(self,name):
        self.name = name
    def kick(self):
        print('I am %s, who call me ' % self.name)
```

```python
a = Ball()
a.setName('dd')
a.kick()
```

## 魔法方法

```python
class Potato:
    def __init__(self,name):
        self.name = name
    def kick(self):
        print('I am %s, who call me ' % self.name)
p = Potato('yy')
p.kick()
```

公有属性和私有属性

```python
#公有
class Person:
    name = 'dd'
p = Person()
p.name
```

```python
#私有
class Person:
    __name = 'dd'
p = Person()
p.name
```

```python
#私有
class Person:
    __name = 'dd'
p = Person()
p._Person__name
```

```python
class Person:
    __name = 'dd'

    def get_name(self):
        return self.__name

p = Person()
print(p.get_name())  # 输出：'dd'
```

继承

```python
class Parent:
    def hello(self):
        print('正在调用父类的方法...')
class Child(Parent):
    pass
p = Parent()
p.hello()
c = Child()
c.hello()
```

```python
import random as r
class Fish:
    def __init__(self):
        self.x =r.randint(0,10)
        self.y =r.randint(0,10)
    def move(self):
        self.x -= 1
        print('我的位置是:',self.x,self.y)
class Goldfish(Fish):
    pass
class Carp(Fish):
    pass
class Salmon(Fish):
    pass
class Shark(Fish):
    def __init__(self):
        self.hungry = True
    def eat(self):
        if self.hungry:
            print('天天吃饭')
            self.hungry = False
        else:
            print('吃不下了')
            
```

```python
fish = Fish()
fish.move()
```

```python
goldfish = Goldfish()
goldfish.move()
```

```python
shark = Shark()
shark.eat()
shark.eat()
```

```python
shark.move()
```

调用未绑定的父类方法

```python
class Shark(Fish):
    def __init__(self):
        Fish.__init__(self)
        self.hungry = True
shark = Shark()
shark.move()
    
```

super函数

```python
class Shark(Fish):
    def __init__(self):
        super().__init__()
        self.hungry = True
        
shark = Shark()
shark.move()
```

多重继承

```python
class Base1:
    def fool(self):
        print('我是fool，我在Base1中.....')
class Base2:
    def foo2(self):
        print('我是foo2，我在Base2中.....')
class Base3:
    def foo3(self):
        print('我是foo3，我在Base3中.....')
class C(Base1,Base2,Base3):
    pass
c = C()
c.fool()
c.foo2()
c.foo3()
```

组合

```python
class Turtle:
    def __init__(self,x):
        self.num = x
class Fish:
    def __init__(self,x):
        self.num = x
class Pool:
    def __init__(self,x,y):
        self.turtle = Turtle(x)
        self.fish = Fish(y)
    def print_num(self):
        print('水里有乌龟%d只，小鱼%d只' %(self.turtle.num,self.fish.num))
```

```python
pool = Pool(1,10)
pool.print_num()
```

```python
class C:
    count = 0
a = C()
b = C()
c = C()
print(a.count,b.count,c.count)
c.count +=10
print(a.count,b.count,c.count)
C.count +=10
print(a.count,b.count,c.count)
```

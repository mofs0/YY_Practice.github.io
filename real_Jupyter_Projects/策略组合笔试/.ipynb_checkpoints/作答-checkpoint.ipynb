```python
import numpy as np
import pandas as pd
```

```python
#定义一个导入函数
def read_stock_data(file_path):
    try:
        df = pd.read_csv(file_path, encoding='gbk', skiprows=1,
                         usecols=['股票代码', '股票名称', '交易日期', '成交额'],
                         index_col=['交易日期'])
        df.index = pd.to_datetime(df.index)
        return df
    except FileNotFoundError:
        print(f"文件 {file_path} 未找到，请检查文件路径。")
    except Exception as e:
        print(f"读取文件 {file_path} 时出现错误: {e}")
```

```python
# 文件列表
file_paths = [
    'sh600439.csv', 'sh600622.csv', 'sh600653.csv',
    'sh603130.csv', 'sh603759.csv', 'sh603933.csv',
    'sz002047.csv', 'sz300050.csv', 'sz301428.csv',
    'sz301520.csv'
]
```

```python
# 读取所有文件
dfs = [read_stock_data(file_path) for file_path in file_paths if read_stock_data(file_path) is not None]
print(dfs)
```

```python
# 按交易时间合并成一个 DF
df = pd.concat(dfs, axis=0)
# 对合并后的DF按索引排序
merged_df = merged_df.sort_index()
print(merged_df)
```

```python
#检查数据
df['成交额'].isnull().any()
```

```python
# 定义查找成交额最的股票的函数
def get_max_volume_stock(group):
    if len(group) == 1:
        return group
    else:
        return group[group['成交额'] == group['成交额'].max()]
```

```python
#输出结果
result_df = merged_df.groupby(level='交易日期').apply(get_max_volume_stock)
result_df = result_df.reset_index(level=0, drop=True)
print(result_df)
```

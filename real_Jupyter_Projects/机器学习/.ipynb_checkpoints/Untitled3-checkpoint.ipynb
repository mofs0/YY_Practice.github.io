```python
#方法二
num = int(input('请输入一个数字'))
def Dec_to_Oct(num):
    result = ""
    while num != 0:
        ret = num % 8
        num //= 8
        result = str(ret) + result
    return result


print('您输入的10进制数其8进制对应的数字为:',Dec_to_Oct(num))
```

第二题

```python
def longest_repeated_substring(s):
    n = len(s)
    longest_substring = ""

    for i in range(2, n + 1, 2):  # 从长度为2的子串开始，逐渐增加长度
        for j in range(n - i + 1):
            substring = s[j:j + i]
            if substring[:i // 2] == substring[i // 2:]:
                if len(substring) > len(longest_substring):
                    longest_substring = substring

    return longest_substring

# 测试示例
input_str = "abababcdabcd"
result = longest_repeated_substring(input_str)
print("最长的自重复子串为:", result)
```

第三题

```python
#要计算无向图的连通分量数量，可以使用深度优先搜索（DFS）算法。
#首先，你需要遍历图中的每个节点，对于每个尚未访问过的节点，进行一次深度优先搜索，并标记所有与该节点连通的节点。
#每次搜索完一个连通分量后，将连通分量的数量加一。
def dfs(graph, visited, node):
    visited[node] = True
    for neighbor in range(len(graph[node])):
        if graph[node][neighbor] == 1 and not visited[neighbor]:
            dfs(graph, visited, neighbor)

def count_connected_components(graph):
    n = len(graph)
    visited = [False] * n
    count = 0

    for i in range(n):
        if not visited[i]:
            dfs(graph, visited, i)
            count += 1

    return count

# 输入
n = int(input("请输入图的节点个数："))
print("请输入邻接矩阵A：")
graph = []
for _ in range(n):
    row = list(map(int, input().split()))
    graph.append(row)

# 输出
result = count_connected_components(graph)
print("该图的连通分量个数为:", result)
```

第四题

```python
from typing import List
import operator
from functools import reduce
class Solution:
    def minFlips(self, mat: List[List[int]]) -> []:
        r, c = len(mat), len(mat[0])
        start = reduce(operator.ior, [t << i * c + j for i, row in enumerate(mat) for j, t in enumerate(row)])
        
        q = [(start, 0)]
        seen = set([start])
        while q:
            cur, step = q.pop(0)
            if not cur:
                return step
            for i in range(r):
                for j in range(c):
                    ne = cur
                    for x, y in (i, j), (i, j + 1), (i, j - 1), (i + 1, j), (i - 1, j):
                        if x >= 0 and x < r and y >= 0 and y < c:
                            ne ^= 1 << x * c + y
                    if ne not in seen:
                        seen.add(ne)
                        q.append((ne, step + 1))
        return -1
solution = Solution()

# 调用 minFlips 方法并传入一个二维列表作为参数
mat = [
    [0, 0,1,0,0,0],
    [0, 1,1,1,0,1],
    [0, 0,1,0,1,1]
]
result = solution.minFlips(mat)

# 打印结果
print("Minimum number of flips required:", result)
```

```python
from typing import List
import operator
from functools import reduce

class Solution:
    def minFlips(self, mat: List[List[int]]) -> int:
        r, c = len(mat), len(mat[0])
        start = reduce(operator.ior, [t << i * c + j for i, row in enumerate(mat) for j, t in enumerate(row)])
        
        q = [(start, 0, set())]  # 添加一个集合来记录翻转的位置
        seen = set([start])
        while q:
            cur, step, flipped = q.pop(0)  # 添加一个变量来记录翻转的位置
            if not cur:
                # 打印翻转位置
                self.printFlippedPositions(flipped, mat)
                return step
            for i in range(r):
                for j in range(c):
                    ne = cur
                    for x, y in (i, j), (i, j + 1), (i, j - 1), (i + 1, j), (i - 1, j):
                        if x >= 0 and x < r and y >= 0 and y < c:
                            ne ^= 1 << x * c + y
                            flipped_pos = (x, y)  # 记录发生翻转的位置
                            flipped.add(flipped_pos)  # 将翻转的位置添加到集合中
                    if ne not in seen:
                        seen.add(ne)
                        q.append((ne, step + 1, flipped.copy()))  # 传递一个副本以避免共享引用
                        flipped.remove(flipped_pos)  # 回退到上一个状态以准备下一次翻转
        return -1
    
    def printFlippedPositions(self, flipped, mat):
        r, c = len(mat), len(mat[0])
        flipped_mat = [[0] * c for _ in range(r)]
        for x, y in flipped:  # 使用记录的翻转位置来更新矩阵
            flipped_mat[x][y] = 1 - mat[x][y]
        print("Flipped positions:")
        for row in flipped_mat:
            print(row)

solution = Solution()

# 调用 minFlips 方法并传入一个二维列表作为参数
mat = [
    [0, 0, 1, 0, 0, 0],
    [0, 1, 1, 1, 0, 1],
    [0, 0, 1, 0, 1, 1]
]
result = solution.minFlips(mat)

# 打印结果
print("Minimum number of flips required:", result)
```

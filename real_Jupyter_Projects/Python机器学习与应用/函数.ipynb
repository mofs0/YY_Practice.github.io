# 函数

```python
#函数定义
def jc(n):
    s = 1
    for i in range(1,n+1):
        s *= i
    return s
```

```python
a = 3
jc(a)
```

```python
def printChars(ch1,ch2,number):
    count = 0
    for i in range(ord(ch1),ord(ch2)+1):
        count +=1
        if count % number != 0:
            print('%4s' % chr(i),end='')
        else:
            print('%4s' % chr(i))
        
```

```python
printChars('!','9',10)
```

```python
def sayHello(s = 'Hello',n=2,m=1):
    for i in range(n):
        print(s * m)
    
```

```python
sayHello()
```

```python
sayHello('HHH',5,2)
```

## 基本数据类型

#### decimal 提供了任意精度的可能

```python
import decimal
from decimal import Decimal
```

```python
decimal.getcontext()
```

```python
d = Decimal(1) / Decimal(11)
d
```

```python
#修改精度
decimal.getcontext().prec = 4
e = Decimal(1) / Decimal(11)
e
```

### 正则表达式

```python
import re
```

```python
series = '''
'01/18/2014 13:00:00',100,'1st';
'01/18/2014 13:30:00',100,'2nd';
'01/18/2014 14:00:00',100,'3rd';
'''
```

```python
dt = re.compile("'[0-9/:\s]+'") #datetime
```

```python
result = dt.findall(series)
result
```

```python
for i in range(1,10):
    if i % 2 == 0:
        print('%d is even' % i)
    elif i % 3 == 0:
        print('%d is multiple of 3' % i)
    else:
        print('%d is odd' % i)
        
```

### 函数式编程

```python
def f(x):
    return x ** 2
f(2)
```

```python
def even(x):
    return x % 2 == 0
even(3)
```

```python
list(map(even,range(10)))
```

```python
list(map(lambda x : x **2,range(10)))
```

```python
list(filter(even,range(15)))
```

### 集合

```python
#每个元素只有一个
s = set(['a','b','c','d','e','f','g'])
s
```

```python
t = set(['a','b','c','t','h','v'])
t
```

```python
#并集
s.union(t)
```

```python
#交集
s.intersection(t)
```

```python
#导入库包
import pandas as pd
#写入数据
data = {
    'month': ['202401', '202402', '202403', '202401', '202402', 
              '202401', '202402', '202403', '202401', '202402', 
              '202403', '202404', '202403'],
    'SKU': ['个护家清', '个护家清', '个护家清', '生鲜', '生鲜', 
            '日用百货', '日用百货', '日用百货', '副食粮油', '副食粮油', 
            '副食粮油', '副食粮油', '日用百货'],
    'GMV': [18108, 3245, 3204, 11843, 14442, 
            8962, 10766, 9927, 3867, 27995, 
            14769, 5675, 9927],
    'Sale': [8031, 1270, 2759, 4975, 7281, 
             3793, 4183, 4487, 2013, 12458, 
             6367, 1850, 4487]
}
df = pd.DataFrame(data)
df
#进行计算
#按月份分组计算GMV总和
monthly_gmv = df.groupby('month')['GMV'].sum().reset_index()
#计算月环比并保留两位小数
monthly_gmv['月环比'] = monthly_gmv['GMV'].pct_change().map(lambda x: f"{x:.2%}" if pd.notna(x) else '—')
#计算年累计
monthly_gmv['年累计'] = monthly_gmv['GMV'].cumsum()
#重命名
monthly_gmv.rename(columns={'month': '月', 'GMV': 'GMV'}, inplace=True)
#调整列顺序
monthly_gmv = monthly_gmv[['月', 'GMV', '月环比', '年累计']]
#输出结果
print(monthly_gmv)
```

```python
from selenium import webdriver
```

```python
driver = webdriver.Chrome()
```

```python
url = 'https://passport.xinpianchang.com/login?redirect_uri=https%3A%2F%2Fwww.xinpianchang.com%2Fuser%2F'
```

```python
driver.get(url)
```

读取文件中的cookie

```python
with open('cookies.text','r',encoding='utf-8') as file:
    cookies_str = file.read()
```

```python
cookies_str
```

使用json将字符串转化

```python
import json
```

```python
cookies = json.loads(cookies_str)
```

向浏览器注入cookie

```python
for cookie in cookies:
    try:
        driver.add_cookie(cookie)
    except Exception as e:
        print(e)
```

```python
driver.get(url)
```

```python
driver.close()
```

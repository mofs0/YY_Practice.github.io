```python
from selenium import webdriver
```

```python
driver = webdriver.Chrome()
```

```python
url = 'https://passport.xinpianchang.com/login?redirect_uri=%2Fsettings%2Fsecurity'
```

```python
driver.get(url)
```

```python
#提取cookie
cookie = driver.get_cookies()
cookie
```

```python
#将数据进行json序列化，转化成文本进行存储
import json
```

```python
cookie_str = json.dumps(cookie)
```

```python
#使用文件存储将字符串存入文件
with open('cookies.text','w',encoding='utf-8') as file:
    file.write(cookie_str)
```

```python
# 读取股票数据
data = pd.read_csv(r'C:/Users/86157/Desktop/茅台2023年股价.csv', parse_dates=True, index_col=0)
data.columns = ['Open', 'High', 'Low', 'Close','Volume','Amount'] 
#计算股票价格的中心趋势
price_mean = data['Close'].mean()
price_median = data['Close'].median()
#计算股票价格的离散与变异程度
price_std = data['Close'].std()
price_var = data['Close'].var()
#计算交易量的分布
volume_max = data['Volume'].max()
volume_min = data['Volume'].min()
volume_quantile = data['Volume'].quantile([0.25,0.5,0.75])
#输出
print('股票价格中心趋势:')
print('平均值:',price_mean)
print('中位数:',price_median)
print('\n股票价格的变异程度:')
print('标准差:',price_std)
print('方差:',price_var)
print('\n股票交易量的分布情况:')
print('最大值:',volume_max)
print('最小值:',volume_min)
print('分位数:',volume_quantile)
```

## 相关性分析

```python
import pandas as pd
import numpy as np
# 1. 原始数据创建
data = {
    'Date': ['2023.1.1', '2023.1.2', '2023.1.3', '2023.1.4', '2023.1.5'],
    'Stock_A': [100, 105, 110, 115, 120],
    'Stock_B': [200, 210, 190, 205, 215]
}

# 2. 创建 DataFrame 并转换索引为 datetime 类型
df = pd.DataFrame(data)
# 关键步骤：将 Date 列转为 datetime 类型，再设为索引（同时删除原 Date 列）
df['Date'] = pd.to_datetime(df['Date'])  # 字符串 → datetime 类型
df = df.set_index('Date').drop(columns='Date', errors='ignore')  # 设为索引并清理重复列
#计算相关性系数
correlation_matrix = np.corrcoef(df['Stock_A'],df['Stock_B'])
#提取系数
correlation = correlation_matrix[0,1]
print('皮尔逊相关系数:',correlation)
```

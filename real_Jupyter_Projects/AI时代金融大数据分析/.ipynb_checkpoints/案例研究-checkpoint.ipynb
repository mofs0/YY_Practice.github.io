```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

# 读取数据并倒序排列
data = pd.read_excel('C:/Users/86157/Desktop/人民币汇率中间价历史数据.xlsx')
data = data[::-1]  # 倒序排列

# 选择前两列并命名
data = data.iloc[:, 0:2]
data.columns = ['Date', 'USD/CNY']

# 删除前两行
data = data.drop(data.head(2).index)

# 转换为DataFrame并设置日期索引
df = pd.DataFrame(data)
df.index = pd.to_datetime(df['Date'])  # 直接将Date列转换为 datetime 索引

# 删除Date列（因为已经作为索引存在）
df = df.drop('Date', axis=1)  # axis=1 表示删除列

# 确保USD/CNY列是数值类型
df['USD/CNY'] = pd.to_numeric(df['USD/CNY'], errors='coerce')
# 处理可能的缺失值
df = df.dropna()

# 划分训练集和测试集
train_data = df.iloc[:-10]
test_data = df.iloc[-10:]

# 拟合ARIMA模型
model = ARIMA(train_data, order=(1,1,1))
model_fit = model.fit()  # 修正了这里的语法错误

print('模型拟合完成')
# 可以查看模型摘要
print(model_fit.summary())
```

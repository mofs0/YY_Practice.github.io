```python
import pandas as pd  # 先确保导入pandas（很多新手会漏这行）
df = pd.read_csv('C:\\Users\\86157\\Desktop\\实例股票数据.csv', encoding='gbk')
print(df)
```

```python
#查看缺少值
missing_values = df.isna().sum()
print('缺失值的数量:',missing_values)
```

### 填充缺失值

```python
#使用平均值填充收盘价
mean_close  = df['收盘价'].mean()
print(mean_close)
df['收盘价'].fillna(mean_close,inplace=True)
print(df['收盘价'])
```

```python
#使用中位数填充最低价
median_low = df['最低价'].median()
print(median_low)
df['最低价'].fillna(median_low,inplace=True)
print(df['最低价'])
```

```python
#用临近值填充最高价
df['最高价'].fillna(method='ffill',inplace=True)
print(df['最高价'])
```

```python
#用0填充成交量
df['成交量'].fillna(0,inplace=True)
print(df['成交量'])
```

```python
#查看填充完成的DF
df
```

### 修改数据类型

```python
#日期格式转换
df['日期'] =pd.to_datetime(df['日期'])
df['日期'].dtype
```

```python
#转换为数值类型
numeric_cols = ['开盘价','收盘价','最高价','最低价','成交量']
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric)
df
```

```python
#数值利列数据类型转
df['开盘价']  =df['开盘价'].astype(float)
df['开盘价'].dtype
```

### 异常值处理

```python
# 步骤1：先计算“成交量为负”的数据（先赋值，再判断，避免提前引用）
negative_volume = df[df['成交量'] < 0]

# 步骤2：正确的if判断格式（冒号+缩进）
if not negative_volume.empty:  # 现在negative_volume已定义，可正常判断是否为空
    print('存在异常值,成交量为负数')
    print(negative_volume)  # 缩进：这行属于if分支，必须缩进

# 步骤3：将负数成交量改为None（这步逻辑没问题，可保留）
df.loc[df['成交量'] < 0, '成交量'] = None  # 注意：None在pandas中会被识别为缺失值

# 步骤4：查看处理后的数据（建议增加“缺失值数量”的打印，验证处理效果）
print('处理后的DF:')
print(df)
print('成交量列的缺失值数量:', df['成交量'].isnull().sum())  # 可选：验证负数是否已转为None
```

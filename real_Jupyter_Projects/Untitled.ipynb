```python
import os
from pdf2docx import Converter
from tqdm import tqdm  # 用于显示进度条，可选

def batch_pdf_to_word(input_dir, output_dir):
    # 确保输出目录存在
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    # 获取所有PDF文件
    pdf_files = [f for f in os.listdir(input_dir) if f.endswith('.pdf')]
    
    if not pdf_files:
        print("目录中没有找到PDF文件")
        return
    
    # 批量转换
    for pdf_file in tqdm(pdf_files, desc="转换进度"):
        pdf_path = os.path.join(input_dir, pdf_file)
        docx_file = pdf_file.replace('.pdf', '.docx')
        docx_path = os.path.join(output_dir, docx_file)
        
        try:
            cv = Converter(pdf_path)
            cv.convert(docx_path)
            cv.close()
            print(f"成功转换：{pdf_file} -> {docx_file}")
        except Exception as e:
            print(f"转换{pdf_file}时出错：{str(e)}")

if __name__ == "__main__":
    # 请修改为你的PDF目录和输出目录
    input_directory = r"C:\Users\86157\Desktop\课程论文\中级微观经济学\黄土高原县域数据汇总\榆林市统计年鉴"  # 你的PDF文件所在目录
    output_directory = r"C:\Users\86157\Desktop\课程论文\中级微观经济学\黄土高原县域数据汇总\榆林市统计年鉴\Word转换结果"  # 转换后的Word保存目录
    
    batch_pdf_to_word(input_directory, output_directory)
    print("批量转换完成！")
```

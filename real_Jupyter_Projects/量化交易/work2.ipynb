```python
import quandl
import pandas as pd
import numpy as np
symbol = 'WIKI/GOOG'
df1 = quandl.get(symbol,start_date='2011-01-01',end_date='2016-12-31')
df2 = quandl.get('WIKI/AAPL',start_date='2011-01-01',end_date='2016-12-31')
df3 = quandl.get('WIKI/FB',start_date='2011-01-01',end_date='2016-12-31')
df1 = pd.DataFrame(df1,columns=['Close'])
df2 = pd.DataFrame(df2,columns=['Close'])
df3 = pd.DataFrame(df3,columns=['Close'])
df1.columns=['GOOG']
df2.columns=['AAPL']
df3.columns=['FB']
exl = pd.concat([df1,df2,df3],axis=1)
exl
```

```python
type(exl)
exl = exl.fillna(exl.mean(axis=0))
exl
```

```python
import matplotlib.pylab as plt
import statsmodels.api as sm
exl['intercept'] = 1
```

```python
X = exl[['intercept', 'AAPL', 'FB']]
y = exl['GOOG']

# 拟合多元线性回归模型
model = sm.OLS(y, X).fit()

# 获取回归结果
print(model.summary())
```

```python
df1 = quandl.get(symbol,start_date='2017-01-01',end_date='2018-12-31')
df2 = quandl.get('WIKI/AAPL',start_date='2017-01-01',end_date='2018-12-31')
df3 = quandl.get('WIKI/FB',start_date='2017-01-01',end_date='2018-12-31')
df1 = pd.DataFrame(df1,columns=['Close'])
df2 = pd.DataFrame(df2,columns=['Close'])
df3 = pd.DataFrame(df3,columns=['Close'])
df1.columns=['GOOG']
df2.columns=['AAPL']
df3.columns=['FB']
new_exl = pd.concat([df1,df2,df3],axis=1)
new_exl
```

```python
new_exl = pd.DataFrame(new_exl)
new_exl['intercept'] = 1
new_exl
```

```python
predictions = model.predict(new_exl)

print(predictions)
```

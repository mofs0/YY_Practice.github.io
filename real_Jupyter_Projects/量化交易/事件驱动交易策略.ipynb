# 事件驱动
为了将事件驱动方法应用于回测系统，有必要定义组件或者对象来处理特定的任务：
Event-事件是事件驱动的基本单元，它包含一种类型(如市场，信号，订单或者填充)，决定如何，将在事件循环内处理
Event Queue 事件队列是存储全部内存的python对象由其他软件生产的事件子类对象
DataHandler - DataHandler是一个提供接口的抽象基类（ABC)用于处理历史或者实时市场数据。这提供了显着的灵活性，在战略和投资组合上灵活使用，该DataHandler会在系统的每个hit line 上产生新的Marketevent
Strategy 该策略也是一种提供市场数据接口ABC并生成相应的SignalEvent，最终由Portfolio使用。一个SignalEvent包含股票代码，一个方向(长期或短期）和一个时间stamp。
Portfolio 这是处理相关订单管理的类层次结构与当前和随后的战略位置。它也执行风险管理整个投资组合，包括部门风险和头寸规模。投资组合需要来自队列的SignalEvents并生成添加到队列的OrderEvent
ExecutionHandler 模拟与经纪商的连接。该处理程序的任务是从队列中接受OrderEvent并通过一个.txt文件执行他们模拟方法或者经纪人的实际连接。一旦执行订单处理程序创建FillEvents，它描述实际进行的实务，包括费用佣金和滑点
Backtest 所有这些组件都包装在正确处理的事件循环中，将它们引向到适当的组件。

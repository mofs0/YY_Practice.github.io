### 输出每列缺失值的具体信息

```python
#法一 根据列名遍历,直接打印空缺值的行
for name in data.columns:
    print('{}列,空缺值的行列{}'.format(name,data[data[name].isnull()].index.tolist()))
```

```python
#法二 标准答案
for name in data.columns:
    if data[name].isnull().any():
        location = data[data[name].isnull() == True].index.to_list()
    print('列名:"{}",第{}行位置有缺失值'.format(name,location))
```

## 基础题

```python
import numpy as np
import pandas as pd
```

### 将下列的字典创建为DataFrame

```python
data = {
    'grammer' : ['Python','C','Java','GO',np.nan,'SOL','PHP','Python'],
    'score' : [1,2,np.nan,4,5,6,7,10]
}
```

```python
df = pd.DataFrame(data)
df
```

### 提取含有字符串’Python‘的行

```python
#方法一
df[df['grammer'] == 'Python']
```

```python
#方法二
results = df['grammer'].str.contains('Python')
results.fillna(value=False,inplace=True)
df[results]
```

### 输出df所有的列名

```python
df.columns
```

### 修改第二列列名为’popularity‘

```python
df.rename(columns= {'score':'popularity'},inplace=True)
df
```

### 统计grammer列中每种汇编语言出现的次数

```python
df.grammer.value_counts()
```

### 将空值用上下值的平均值填充

```python
df1 = df[:].copy()
#发现null值,返回逻辑值
null_values = df1['popularity'].isnull()
for i in range(1,len(df1)-1): #因为是上下值填充所有需要限制范围
    if null_values[i]:#i=True表示空值,需要填充
        avg = (df1.at[i-1,'popularity'] + df1.at[i+1,'popularity']) / 2
        df1.at[i,'popularity'] = avg
print(df1)
```

```python
#简单方法
df['popularity'] = df['popularity'].fillna(df['popularity'].interpolate())
df
```

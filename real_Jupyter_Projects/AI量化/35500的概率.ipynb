```python
import math
from scipy.stats import norm

# 参数设置
n = 10000  # 掷骰子的次数
mu = 3.5  # 骰子点数的平均值
sigma = math.sqrt(35/12)  # 骰子点数的标准差

# 计算标准差
std_deviation = math.sqrt(n) * sigma

# 计算Z-score
z = (35500 - n * mu) / std_deviation

# 计算概率
probability = 1 - norm.cdf(z)

print("总点数超过35500的概率为:", probability)
```

# 股票买卖收益分析

```python
#策略:每个月的第一个交易日买入一手，买一年最后一起卖出
df
```

```python
new_df = df['2010-01':'2020-02']
new_df
```

```python
#买股票
df_monthly = new_df.resample('M').first()#根据月份从原始数据中提取指定的数据
df_monthly
```

```python
#买入股票的花费
cos =df_monthly['open'].sum()*100
cos
```

```python
#卖出股票的钱
#切除最后一年
df_yearly = new_df.resample('A').last()[:-1]
resv = df_yearly['open'].sum()*1200
resv
```

```python
#估计手中剩余的股票计为收益
#使用昨日收盘价作为单价
last_monry = 200 * new_df['close'][-1]
last_monry
```

```python
#计算总收益
resv+last_monry-cos
```

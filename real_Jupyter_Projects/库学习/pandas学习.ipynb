# 数据抽取
DataFrame对象中的loc属性和iloc属性
loc属性 ：以列名和行名为参数，当只有一个参数时候，默认是行名，即抽取整行的数据，包括所有列。
iloc属性：以行和列位置为索引(即0，1，2)作为参数，0表示第一行，1表示第二行。当只有一个参数时，默认行索引，即抽取整行数据包括所有列。

# 索引的设置
pandas索引的作用 方便查数据 使用索引可以提升查询性能

数据清洗后重新设置连续的索引 df.reset_index()

```python
s1=pd.Series([10,20,30],index=['a','b','c'])
s2=pd.Series([2,3,4],index=['a','b','c'])
s=s1+s2
s
```

```python
#Series重新设置索引
s=pd.Series([11,22,33],index=[1,2,3])
#重新设置索引
print(s.reindex(range(1,6)))#多出来2个nan
#使用0进行填充
print(s.reindex(range(1,6),fill_value=0))
#向前填充和向后填充
print(s.reindex(range(1,6),method='ffill'))#向前填充
print(s.reindex(range(1,6),method='bfill'))#向后填充
```

# 数据计算
求和 sum
求均值 mean
最大值 max
最小值 min
中位数 media
众数 mode
方差 var
标准差 std
分为数 quantile

```python
#行计算
data =[[100,77,88],[90,87,88],[67,77,99]]
columns = ['数学','语文','英语']
df = pd.DataFrame(data,columns=columns,index=[1,2,3])
df['总成绩']=df.sum(axis=1)
df['平均分'] = df[['语文', '数学', '英语']].mean(axis=1)
```

```python
df
```

```python
#列计算
df=df[['语文', '数学', '英语']]._append(df.mean(),ignore_index=True)
df
```

```python
#众数 列和行
data =[[100,77,88],[90,87,88],[90,77,99]]
columns = ['数学','语文','英语']
df = pd.DataFrame(data=data,columns=columns,index=[1,2,3])
df.mode()
print(df.mode())
print(df.mode(axis=1))
print(df['数学'].mode())
```

```python
#求方差
data = [[135,140,138,145,140],[140,145,139,142,132]]
index = ['李帆','张曙光']
columns=['一模','二模','三模','四模','五模']
df=pd.DataFrame(data,index,columns)
df.var(axis=1)
```

```python
#求标准差
data = [[135,140,138,145,140],[140,145,139,142,132]]
index = ['李帆','张曙光']
columns=['一模','二模','三模','四模','五模']
df=pd.DataFrame(data,index,columns)
df.std(axis=0)
```

```python
#求分位数
data = [120,110,112,100,98,34,123,115]
columns = ['数学']
df = pd.DataFrame(data=data,columns=columns)
x=df['数学'].quantile(0.35)
print(x)
print(df[df['数学']<x])
```

数据表的拼接

```python
left = pd.DataFrame({'key1':['k0','k0','k1','k2'],
                    'key2':['k0','k1','k0','k1'],
                    'A':['A0','A1','A2','A3'],
                    'B':['B0','B1','B2','B3']})
right = pd.DataFrame({'key1':['k0','k1','k1','k2'],
                    'key2':['k0','k0','k0','k0'],
                    'C':['C0','C1','C2','C3'],
                    'D':['D0','D1','D2','D3']})
```

```python
pd.merge(left,right,on=['key1','key2'])
```

```python
pd.merge(left,right,on=['key1','key2'],how='inner')
```

```python
pd.merge(left,right,on=['key1','key2'],how='left')
```

```python
pd.merge(left,right,on=['key1','key2'],how='right')
```

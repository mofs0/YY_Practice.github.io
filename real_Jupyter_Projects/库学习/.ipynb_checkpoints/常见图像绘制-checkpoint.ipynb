```python
import matplotlib.pyplot as plt
import random
import numpy as np
from pylab import mpl
mpl.rcParams['font.sans-serif'] = ['SimHei']
mpl.rcParams['axes.unicode_minus'] = False
```

```python
#散点图
x = [random.randint(1,1000) for i in range(1000)]
y = [random.randint(1,1000) for i in range(1000)]
```

```python
plt.figure(figsize=(20,8),dpi=100)
plt.scatter(x,y)
plt.show()
```

```python
#柱状图
movie_name = ['雷神3:诸神黄昏','正义联盟','东方快车谋杀案','寻梦环游记','全球风暴','降魔传','追捕']
x=range(len(movie_name))
y = [123,345,454,333,666,943,546]
plt.figure(figsize=(20,8),dpi=100)
plt.bar(x,y,width=0.5,color=['b','r','y','g','b','r','c'])
plt.xticks(x,movie_name)
plt.grid(linestyle='--',alpha=0.5)
plt.title('电影收入')
plt.show()
```

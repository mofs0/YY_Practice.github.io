```python
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
np.random.seed(0)#随机种子
arr = np.random.randn(100)
sns.distplot(arr,bins=10,hist=True,kde=True,rug=True)
```

```python
#双变量绘图
dataframe_obj = pd.DataFrame({'x':np.random.randn(500),'y':np.random.randn(500)})
```

```python
sns.jointplot(x='x',y='y',data=dataframe_obj,kind='scatter',color='r',size=10,ratio=5)
```

```python
#绘制二维直方图
sns.jointplot(x='x',y='y',data=dataframe_obj,kind='hex')
```

```python
#绘制核密度估计图像
sns.jointplot(x='x',y='y',data=dataframe_obj,kind='kde',shade=True)
```

```python
#绘制成对变量分布
dataset = sns.load_dataset('iris')
dataset.head()
```

```python
sns.pairplot(data=dataset)
```

散点图

```python
tips = sns.load_dataset('tips')
sns.stripplot(x='day',y='total_bill',data=tips)
```

```python
sns.stripplot(x='day',y='total_bill',data=tips,jitter=False)
```

```python
sns.stripplot(x='day',y='total_bill',data=tips,hue='time',jitter=True)
```

```python
sns.swarmplot(x='day', y='total_bill', data=tips)
```

箱型图

```python
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
sns.boxplot(x='day',y='total_bill',data=tips,hue='time',palette=['g','r'],saturation=0.9)
```

小提琴图

```python
sns.violinplot(x='day',y='total_bill',data=tips)
```
